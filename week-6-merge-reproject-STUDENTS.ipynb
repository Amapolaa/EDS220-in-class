## Merge Data Frames
---
- We can use `country_names` data frame to add the country names into the `n_comms` data which,in turn, will allow us to merge the data frame with the countries data

To merge data frames we can use `pandas.merge()` function

```python
output_df = pd.merge(left_df,
                    right_df,
                    how = type_of_join,
                    on = column_to_join)
where
- output_df is the data frame resulting from the merge,
- left_df is the data frame we have “on the left side”,
- right_df is the data frame we have “on the right side”,
- how specifies the type of join between the left and right data frames, (check -- the options here), the default is to do an inner join,on specifies the column to join on; 
- this column must be present in both our data frames.
```

````python
n_comms = pd.merge(n_comms,
                   country_names,
                   how = 'left',
                   on = 'country')
````

````python
check_common_columns(n_comms, countries)
````

## Check-In
---
- Create a new variable named arctic_countries which is the result of an inner join between our countries and n_comms data frames. The inner join will merge the subset of rows that have matches in both the left table and the right table.

````python
arctic_countries = pd.merge(countries,
                           n_comms,
                           how = 'inner',
                           on = 'admin')

arctic_countries
````

### **Note:** 
---
- when merging with a geoDataframe, the GeoDataFrame needs to be the left to keep the geometry
- wheneveer you merge you are prob going to be loosing data, and need to check if everything else is still working.

### Review results

````python
#lets check the original data frame and then the one in the merged dataframe
print(f"Countries in n_comms: {len(n_comms)}")
print(f"Countries in arctic_countries: {len(arctic_countries)}")
````

````python
#countries in n_comms without a match in countries
# starting in n_comms there are still more data
#
n_comms[~n_comms['admin'].isin(countries['admin'])]
````

````python
#investigating in the countries dataframe to see how Aland Islands looks 
countries[countries['admin'].str.contains('Aland')]
# in admin its aland county n the other says aland islands
````

````python
# were going to fix it manually and match the names then merge
# match the name used in the countries data
n_comms['admin'] = n_comms['admin'].replace('Aland Islands', 'Aland')

#now lets merge again
arctic_countries = countries.merge(n_comms,
                                 how = 'inner',
                                 on = 'admin')
````

````python
print(print(f"Countries in arctic_countries: {len(arctic_countries)}")
)
````

## Reprojecting
---
- Reprojecting means changing the coordinate reference system of your geospatial data.

````python
# Reproject to CRS optimized for Arctic region
arctic_countries = arctic_countries.to_crs('epsg:3413')

arctic_countries.plot()
````

````python
fig, ax = plt.subplots(figsize=(8,6))

#removes the axis for a cleaner map
ax.axis('off')

# Create choropleth map of communities
# Plot with refined color and edge style
arctic_countries.plot(
    ax=ax,
    column='n_communities',
    cmap = 'PuBuGn',
    legend=True,
    edgecolor="0.6",
    linewidth=0.5,
    legend_kwds={
        "shrink": 0.7,
        "label": "Number of Communities",
        "orientation": "horizontal",
        "pad": 0.05
    }
)

# Add title for better context
ax.set_title('Communities at or above 55°N, 2022\n(at least 10,000 people; 500 in Alaska)',
             fontsize=14, weight='bold', pad=15)

plt.show()
````
